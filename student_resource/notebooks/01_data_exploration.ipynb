```python
def audit_source_file(file_path, expected_prefix, chunksize=100_000):
    total_rows = 0
    missing_name = 0
    missing_address = 0
    missing_country = 0
    duplicate_ids = 0
    invalid_prefix = 0
    countries = {}

    seen_ids = set()

    for chunk in pd.read_csv(
        file_path,
        sep="\t",
        chunksize=chunksize,
        usecols=["entity_id", "business_name", "business_address", "country"],
        dtype=str
    ):
        total_rows += len(chunk)

        missing_name += chunk["business_name"].isna().sum()
        missing_address += chunk["business_address"].isna().sum()
        missing_country += chunk["country"].isna().sum()

        # Check IDs
        for entity_id in chunk["entity_id"]:
            if entity_id in seen_ids:
                duplicate_ids += 1
            else:
                seen_ids.add(entity_id)

            if not str(entity_id).startswith(expected_prefix):
                invalid_prefix += 1

        # Country counts
        country_counts = chunk["country"].value_counts(dropna=False)

        for country, count in country_counts.items():
            countries[country] = countries.get(country, 0) + count

    return {
        "rows": total_rows,
        "missing_name": missing_name,
        "missing_address": missing_address,
        "missing_country": missing_country,
        "duplicate_entity_ids": duplicate_ids,
        "invalid_prefix": invalid_prefix,
        "countries": countries
    }
```

```python
source1_audit = audit_source_file(
    TRAIN_DIR / "train_source1.tsv",
    "S1-"
)

source2_audit = audit_source_file(
    TRAIN_DIR / "train_source2.tsv",
    "S2-"
)

source3_audit = audit_source_file(
    TRAIN_DIR / "train_source3.tsv",
    "S3-"
)
```

```python
ground_truth = pd.read_csv(
    TRAIN_DIR / "train_ground_truth.tsv",
    sep="\t",
    dtype=str
)

ground_truth.head()
```

```python
def count_matches(value):
    if pd.isna(value) or str(value).strip() == "":
        return 0

    return len(str(value).split(","))


ground_truth["match_count"] = (
    ground_truth["matched_entity_ids"]
    .fillna("")
    .apply(count_matches)
)
```

```python
ground_truth["match_count"].describe()
```

```python
ground_truth["match_count"].value_counts().sort_index()
```

```python
singleton_count = (ground_truth["match_count"] == 0).sum()
total_entities = len(ground_truth)

print("Total Source 1 entities:", total_entities)
print("Singletons:", singleton_count)
print(
    "Singleton percentage:",
    round(singleton_count / total_entities * 100, 2),
    "%"
)
```

```python
def count_source_matches(value):
    if pd.isna(value) or str(value).strip() == "":
        return 0, 0

    ids = str(value).split(",")

    s2 = sum(entity_id.startswith("S2-") for entity_id in ids)
    s3 = sum(entity_id.startswith("S3-") for entity_id in ids)

    return s2, s3


ground_truth[["s2_count", "s3_count"]] = (
    ground_truth["matched_entity_ids"]
    .fillna("")
    .apply(
        lambda x: pd.Series(count_source_matches(x))
    )
)
```

```python
print("Total S2 matches:", ground_truth["s2_count"].sum())
print("Total S3 matches:", ground_truth["s3_count"].sum())
```
